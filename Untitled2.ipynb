```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# SETUP & DISTANCE MATRIX
# ==========================================
np.random.seed(42)
n_cities = 20
n_ants = 30
population_size = 50
iterations = 100

cities = np.random.rand(n_cities, 2) * 100
distance = np.zeros((n_cities, n_cities))
for i in range(n_cities):
    for j in range(n_cities):
        distance[i, j] = np.linalg.norm(cities[i] - cities[j])

def route_length(route):
    total = 0
    for i in range(len(route)):
        total += distance[route[i], route[(i + 1) % len(route)]]
    return total

# ==========================================
# 1. ANT COLONY OPTIMIZATION (ACO)
# ==========================================
def run_aco():
    alpha, beta, evaporation, Q = 1.0, 5.0, 0.5, 100.0
    pheromone = np.ones((n_cities, n_cities))
    best_distance = float("inf")
    convergence = []

    for _ in range(iterations):
        all_routes, all_distances = [], []
        for _ in range(n_ants):
            start = np.random.randint(n_cities)
            route = [start]
            unvisited = list(range(n_cities))
            unvisited.remove(start)

            while unvisited:
                current = route[-1]
                probabilities = [
                    (pheromone[current][city] ** alpha) * ((1.0 / (distance[current][city] + 1e-10)) ** beta)
                    for city in unvisited
                ]
                probabilities = np.array(probabilities)
                probabilities /= probabilities.sum()
                next_city = np.random.choice(unvisited, p=probabilities)
                route.append(next_city)
                unvisited.remove(next_city)

            d = route_length(route)
            all_routes.append(route)
            all_distances.append(d)
            if d < best_distance:
                best_distance = d

        pheromone *= (1.0 - evaporation)
        for route, d in zip(all_routes, all_distances):
            for i in range(n_cities):
                a, b = route[i], route[(i + 1) % n_cities]
                pheromone[a][b] += Q / d
                pheromone[b][a] += Q / d
        convergence.append(best_distance)
    return convergence

# ==========================================
# 2. GENETIC ALGORITHM (GA)
# ==========================================
def run_ga():
    population = [np.random.permutation(n_cities) for _ in range(population_size)]
    best_distance = float("inf")
    convergence = []

    for _ in range(iterations):
        scores = np.array([route_length(route) for route in population])
        idx = np.argmin(scores)
        if scores[idx] < best_distance:
            best_distance = scores[idx]

        new_population = [population[idx].copy()]
        while len(new_population) < population_size:
            p1 = population[np.random.randint(population_size)]
            p2 = population[np.random.randint(population_size)]

            a, b = sorted(np.random.choice(n_cities, 2, replace=False))
            child = np.full(n_cities, -1)
            child[a:b+1] = p1[a:b+1]
            remaining = [x for x in p2 if x not in child]
            positions = [i for i in range(n_cities) if child[i] == -1]
            for pos, city in zip(positions, remaining):
                child[pos] = city

            if np.random.rand() < 0.2:
                x, y = np.random.choice(n_cities, 2, replace=False)
                child[x], child[y] = child[y], child[x]

            new_population.append(child)
        population = new_population
        convergence.append(best_distance)
    return convergence

# ==========================================
# RUN BOTH & PLOT COMPARISON ON 1 GRAPH
# ==========================================
aco_curve = run_aco()
ga_curve = run_ga()

print(f"ACO Final Best Distance: {round(aco_curve[-1], 2)}")
print(f"GA Final Best Distance:  {round(ga_curve[-1], 2)}")

plt.figure(figsize=(9, 5))
plt.plot(aco_curve, label="Ant Colony Optimization (ACO)", color="blue", linewidth=2)
plt.plot(ga_curve, label="Genetic Algorithm (GA)", color="red", linewidth=2)
plt.title("Performance Comparison: ACO vs. GA")
plt.xlabel("Number of Iterations")
plt.ylabel("Best Route Distance")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
